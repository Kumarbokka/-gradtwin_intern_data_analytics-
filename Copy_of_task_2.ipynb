1. Write a program to check whether a year is a leap year or not.

```python
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("It is a leap year")
else:
    print("It is not a leap year")
```

4. Check the experience of the user and give a Diwali bonus according to their experience.  

```python
experience = float(input("Enter your experience in years: "))
salary = float(input("Enter your salary: "))

if experience >= 10:
    bonus = salary * 0.20
elif experience >= 5:
    bonus = salary * 0.15
elif experience >= 2:
    bonus = salary * 0.10
else:
    bonus = salary * 0.05

print("Diwali Bonus =", bonus)
print("Total Salary =", salary + bonus)
```

5. Check whether the person wants to take a book from the library.

```python
books = ["Python", "Java", "C", "HTML", "SQL"]

want_book = input("Do you want to take a book? (yes/no): ")

if want_book.lower() == "yes":

    card = input("Do you have a library card? (yes/no): ")

    if card.lower() == "yes":

        print("Available books:")
        print(books)

        book = input("Which book do you want? ")

        if book in books:
            books.remove(book)
            print("Book issued successfully!")
            print("Remaining books:", books)
        else:
            print("Book is not available")

    else:
        print("You need a library card to take a book.")

else:
    print("Okay, no book taken.")
```

1. Write a program to print the factorial of a number.  

```python
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

2. Write a program to check whether the number is Armstrong or not.  

```python
num = int(input("Enter a number: "))

original = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total = total + digit ** digits
    num = num // 10

if total == original:
    print("It is an Armstrong number")
else:
    print("It is not an Armstrong number")
```

4. Write a program to print the sum of digits accepted from the user.  

```python
num = int(input("Enter a number: "))

total = 0

while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10

print("Sum of digits =", total)
```

1. Write a Python function that takes a list and returns a new list with distinct (unique) elements from the first list. Sample List: [1,2,3,3,3,3,4,5]

```python
def unique_list(numbers):
    unique = []

    for num in numbers:
        if num not in unique:
            unique.append(num)

    return unique


numbers = [1, 2, 3, 3, 3, 3, 4, 5]

result = unique_list(numbers)

print("Original List:", numbers)
print("Unique List:", result)
```

2. Write a function count_vowels(string) that takes a string and returns the number of vowels (a, e, i, o, u) in the string.  

```python
def count_vowels(string):
    count = 0

    for char in string:
        if char.lower() in "aeiou":
            count = count + 1

    return count


text = input("Enter a string: ")

result = count_vowels(text)

print("Number of vowels =", result)
```

3. Create a function is_prime(n) that takes a number and returns True if the number is prime, otherwise False.

```python
def is_prime(n):

    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

if is_prime(num):
    print("The number is prime")
else:
    print("The number is not prime")
```

4. Create a simple calculator function.  

```python
def calculator(a, b, operator):

    if operator == "+":
        return a + b

    elif operator == "-":
        return a - b

    elif operator == "*":
        return a * b

    elif operator == "/":
        if b != 0:
            return a / b
        else:
            return "Cannot divide by zero"

    else:
        return "Invalid operator"


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

result = calculator(num1, num2, operator)

print("Result =", result)
```
