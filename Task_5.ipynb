### 19. Count how many even and odd numbers are present in a list.

```python
numbers = [10, 25, 7, 44, 18, 31]
even = 0
odd = 0
for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1
print("Even:", even)
print("Odd:", odd)
```

### 20. Count the number of positive, negative and zero values in a list.

```python
numbers = [10, -5, 0, 8, -2, 0, 15]
positive = 0
negative = 0
zero = 0
for n in numbers:
    if n > 0:
        positive += 1
    elif n < 0:
        negative += 1
    else:
        zero += 1
print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```

### 27. Reverse a given number using a while loop.

```python
n = int(input("Enter a number: "))
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10
print(reverse)
```

### 30. Check whether a number is a palindrome using a while loop.

```python
n = int(input("Enter a number: "))
original = n
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10
if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```
