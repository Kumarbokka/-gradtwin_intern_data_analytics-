Print numbers from 1 to 10 using a for loop.

```python
for i in range(1, 11):
    print(i)
```

Print all even numbers from 1 to 20 using a for loop.

```python
for i in range(2, 21, 2):
    print(i)
```

Print the multiplication table of 5 using a for loop.

```python
for i in range(1, 11):
    print("5 x", i, "=", 5 * i)
```

Print each character in the word "Python" using a for loop.

```python
word = "Python"

for character in word:
    print(character)
```

Print numbers from 1 to 20, but stop the loop when the number reaches 10 using break.

```python
for i in range(1, 21):
    if i == 10:
        break
    print(i)
```

Skip number 10 using continue

```python
for i in range(1, 21):
    if i == 10:
        continue
    print(i)
```

Use pass when the number is 30

```python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        pass
    print(number)
```

Skip even numbers and print only odd numbers

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

Stop when a number greater than 15 is found

```python
numbers = [5, 8, 12, 15, 20, 25]

for number in numbers:
    if number > 15:
        break
    print(number)
```

Names — continue and pass

```python
names = ["Arun", "Bala", "Cathy", "Divya", "Esha"]

for name in names:
    if name == "Cathy":
        continue
    if name == "Divya":
        pass
    print(name)
```

While loop

```python
attempts = 1

while attempts <= 3:
    print("Enter your PIN - Attempt", attempts)
    attempts = attempts + 1

print("Maximum attempts completed")
```
