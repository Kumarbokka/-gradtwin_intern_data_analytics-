## Exception Handling

### Q11. Write a Python program using try and except to handle division by zero.

```python
try:
    a = 10
    b = 0
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### Q12. Write a Python program to divide two numbers and handle ZeroDivisionError.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### Q13. Write a Python program to take an integer as input and handle invalid input using ValueError.

```python
try:
    number = int(input("Enter an integer: "))
    print(number)
except ValueError:
    print("Invalid input")
```

### Q14. Write a Python program using try, except, and finally.

```python
try:
    a = int(input("Enter a number: "))
    print(10 / a)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Program completed")
```

### Q15. Write a Python program that handles an error when converting a string into an integer.

### Q16. Write a Python program to perform addition of two numbers using try-except and handle invalid input.

## Deep Copy

### Q21. Write a Python program to create a deep copy of a list using deepcopy().

```python
import copy
list1 = [1, 2, 3, 4]
list2 = copy.deepcopy(list1)
print(list2)
```

### Q22. Write a Python program to create a nested list and make a deep copy of it.

```python
import copy
list1 = [[1, 2], [3, 4]]
list2 = copy.deepcopy(list1)
print(list1)
print(list2)
```

### Q23. Write a Python program to modify the deep-copied nested list and print both lists.

```python
import copy
list1 = [[1, 2], [3, 4]]
list2 = copy.deepcopy(list1)
list2[0][0] = 10
print(list1)
print(list2)
```

### Q24. Write a Python program that creates both a shallow copy and deep copy of a nested list.

```python
import copy
list1 = [[1, 2], [3, 4]]
shallow_copy = copy.copy(list1)
deep_copy = copy.deepcopy(list1)
print(list1)
print(shallow_copy)
print(deep_copy)
```

### Q25. Write a Python program to create a nested list, make a deep copy, modify the copied list, and verify that the original list is unchanged.

```python
import copy
list1 = [[1, 2], [3, 4]]
list2 = copy.deepcopy(list1)
list2[0][0] = 10
print("Original list:", list1)
print("Deep copied list:", list2)
```
