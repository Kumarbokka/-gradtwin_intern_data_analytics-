What is an Operator?

An operator is a special symbol or keyword used to perform an operation on values or variables.

```python
a = 10
b = 5

print(a + b)
```

Here:

a and b → Operands
+ → Operator
15 → Result
Easy way to remember:

Operand + Operator + Operand = Result

2.Comparison Operators

3.Logical Operators

4.Assignment Operators

1.Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations on numbers.

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a ** b)
```

2.Comparison Operators

Comparison operators are used to compare two values.

```python
marks = 75

print("Equal:", marks == 75)
print("Greater than 50:", marks > 50)
print("Less than 50:", marks < 50)
print("Not equal to 40:", marks != 40)
```

3. Logical Operators

Logical operators are used to combine or reverse multiple conditions.

```python
age = 21
marks = 80

print(age >= 18 and marks >= 50)
print(age < 18 or marks >= 50)
print(not age < 18)
```

4. Assignment Operators

Assignment operators are used to assign or update values in variables

```python
score = 50

score += 10
print("After += :", score)

score -= 5
print("After -= :", score)

score *= 2
print("After *= :", score)

score /= 5
print("After /= :", score)

score %= 3
print("After %= :", score)

score **= 2
print("After **= :", score)
```

5. Bitwise Operators

Bitwise operators work directly on the binary representation (0s and 1s) of integers.

```python
a = 5
b = 3

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

6. Membership Operators

Membership operators are used to check whether a value is present inside a sequence or collection.

```python
subjects = ["Python", "SQL", "Java"]

print("Python" in subjects)
print("HTML" in subjects)
print("Java" not in subjects)
```

7. Identity Operators

Identity operators are used to check whether two variables refer to the same object in memory

```python
a = [10, 20, 30]
b = a
c = [10, 20, 30]

print(a is b)
print(a is c)
print(a is not c)
```
